



 
# use the Titanic dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset("titanic")
df.head()
# sns.scatterplot(x='class',y='gende',data=df,hue='Survival')
# plt.title("Class vs Gende")
# plt.show()

```

# 1.Compare Passenger Survival by Class and Gende

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df, x="pclass", y="survived", hue="sex", palette="Set1")
plt.title("Passenger Survival Rate by Class and Gender")
plt.show()
```

# 2. Distribution of Passenger Ages

```python
sns.histplot(df["age"].dropna(), kde=True, bins=30, color="blue")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

# 3. Fare Distribution by Class

```python

sns.boxplot(data=df, x="pclass", y="fare", palette="coolwarm")
plt.title("Fare Distribution by Class")
plt.show()
```

# 4. Relationship Between Age and Fare

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="age", y="fare", hue="pclass", style="sex", palette="viridis")
plt.title("Relationship Between Age and Fare")
plt.show()

```

# 5. Correlation Heatmap

```python
# 5. Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
