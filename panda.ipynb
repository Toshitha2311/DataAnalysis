#### READING AND EXPLORING THE DATA

```python
#4. Reading & Exploring Data
# Read CSV (replace with your file path)
df = pd.read_csv("/Users/toshitha/Downloads/dfall_clean.csv")
 

# print(df.head())
# print(df.tail())
# print(df.info())
# print(df.describe())
# print(df.shape)
# Example DataFrame
data = {'Name':['Alice','Bob','Charlie'],'Age':[25,30,35],'Sex':['F','M','M']}
print("-----------------")
df = pd.DataFrame(data)
print("-----------------")
 
# Explore
print(df.head())

print("########")
print(df.tail())
print("########")
print(df.info())
print("########\n        \n")
print(df.describe())
print("########")
print(df.shape)
```

#### 5. Selecting Columns & Rows

```python
# Select column
print(df['Name'])
 
# Select multiple columns
print(df[['Name','Age']])
 
# Select row by index
print(df.iloc[0]) 
print("---------------------------------")      # first row
print(df.loc[0])        # first row using label
 
# Select subset of rows and columns
print(df.loc[0:1, ['Name','Sex']])

#6. Filtering & Sorting
# Filter rows
print(df[df['Age'] > 28])
 
# Sort by Age
print(df.sort_values('Age', ascending=False))
```

```python
#. Adding & Removing Columns
# Add new column
df['Age_in_5yrs'] = df['Age'] + 5
print(df)
 
# Drop column
df = df.drop('Age_in_5yrs', axis=1)
print(df)
```

```python
#8. GroupBy & Aggregation
import pandas as pd
data = {'Name':['Alice','Bob','Charlie','Alice','Bob'],
        'Score':[85,90,95,80,70]}
df = pd.DataFrame(data)
print(df)
# Group by Name and calculate mean score
grouped = df.groupby('Name').mean()
print(grouped)
```
