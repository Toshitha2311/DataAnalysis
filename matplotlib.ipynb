# Matplotlib-Python Visualization Library

## 1. What is Matplotlib
  Matplotlib is a python library for creating,static,animated and interactive visualizations.

## Why we need it?
  It helps to understand trends,pattern and outlayers.

## Where we use it?
 1.In Data Analysis

 2.In Machine learning(For plotting lose curves,linear regression)

 3.For Business purposes(for analysis the market trends ),finance and scientific research.




# 2.Import Matplotlib


```python
import matplotlib.pyplot as plb
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,25,30,40]
 
plt.plot(x, y,color='green')
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
categories = ['A','B','C','D']
values = [10,15,7,12]
#plt.plot("x", "y")
plt.xlabel("X-axis")
plt.ylabel("y-axis")
plt.bar(categories, values, color='green')
plt.title("Bar Plot Example")
plt.show()
```

```python
x = [5,7,8,7,2,17,2,9,4,11]
y = [99,86,87,88,100,86,103,87,94,78]
 
plt.scatter(x, y, color='red', marker='*')
plt.title("Scatter Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
data = [22,87,5,43,56,73,55,54,11,20,51,5,79,31,27]
 
plt.hist(data, bins=5, color='orange', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel("Bins")
plt.ylabel("Frequency")
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [2,3,5,7,11]
y2 = [1,4,6,8,10]
 
plt.subplot(1,2,1)  # 1 row, 2 columns, plot 1
plt.plot(x, y1, color='blue')
plt.title("Plot 1")
 
plt.subplot(1,2,2)  # 1 row, 2 columns, plot 2
plt.plot(x, y2, color='red')
plt.title("Plot 2")
 
plt.tight_layout()
plt.show()
 
```
