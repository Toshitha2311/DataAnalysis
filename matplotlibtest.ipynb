# 1.Create a line plot of student marks across 5 exams

```python
import matplotlib.pyplot as plt
data=[45,37,48,34,23]
plt.plot(data,color='violet')
plt.title("Line Plot of Student Marks Across 5 Exams")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

```

# 2.Plot a bar chart of sales of 4 products.

```python
products=['Pen','Pencil','Book','Mobile']
data=[67,45,90,67]
plt.title("Bar chart of sales of 4 products")
plt.xlabel(products)
plt.ylabel("Sales")
plt.bar(products,data,color='grey')
plt.show()
```

# 3.Make a scatter plot of Age vs Salary.

```python
age=[45,65,76,83,65,56,76]
Salary=[45000,56777,87999,98000,98432,87432,67890]
plt.scatter(age,Salary,color='lightgreen',marker='*',alpha=0.8)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Scatter plot of Age vs Salary")
plt.show()
```

# 4.Draw a histogram of marks scored by students in a class.

```python
marks=[56,43,6,83,51,25,65,21]
plt.hist(marks,bins=4,color='Brown',edgecolor='black')
plt.title("Histogram of Marks Scored by Students in a Class")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

# 5.Customize a plot with different colors, markers, and line styles.

```python
import numpy as np
x = np.linspace(0, 10, 100)
y = np.cos(x)
plt.plot(x, y, color='blue', marker='o', linestyle='--', markersize=2, linewidth=2)
plt.title("Customized Sine Wave Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True) 
plt.show()
```

# 6.Create two subplots: Line plot and bar plot side by side.

```python
x = [1,2,3,4,5]
y1 = [12,13,11,17,11]
y2 = [1,4,6,8,10]
 
plt.subplot(1,2,1)  # 1 row, 2 columns, plot 1
plt.plot(x, y1, color='blue')
plt.title("Plot 1")
 
plt.subplot(1,2,2)  # 1 row, 2 columns, plot 2
plt.bar(x, y2, color='red')
plt.title("Plot 2")
 
plt.tight_layout()
plt.show()
```
