```python
import pandas as pd
data = pd.read_csv(
    "/Users/toshitha/Downloads/loan.csv"
)
applicant_mean=data['ApplicantIncome'].mean()
applicant_std=data['ApplicantIncome'].std()
print("Applicant Income Mean:",applicant_mean)
print("Applicant Income Standard Deviation:",applicant_std)
z1=(5849-applicant_mean)/applicant_std
print("Z score for 5849:",z1)
z2=(4583-applicant_mean)/applicant_std
z3=(3000-applicant_mean)/applicant_std
z4=(2583-applicant_mean)/applicant_std
z5=(6000-applicant_mean)/applicant_std
print("Z score for 4583:",z2)
print("Z score for 3000:",z3)
print("Z score for 2583:",z4)       
print("Z score for 6000:",z5)
```

```python
appli_skewnwss=data['ApplicantIncome'].skew()
print(appli_skewnwss)
if(appli_skewnwss>1 or appli_skewnwss<-1):
 print('Highly skewed data')
elif appli_skewnwss<0.5 and appli_skewnwss>-0.5:
  print('Fairly skewed data')
elif appli_skewnwss>=0.5 and appli_skewnwss<=1 or appli_skewnwss<=-0.5 and appli_skewnwss>=-1:
  print('Moderately skewed data')
```

```python
coapp_sk=data['CoapplicantIncome'].skew()
print(coapp_sk)
if(coapp_sk>1 or coapp_sk<-1):
 print('Highly skewed data')
elif coapp_sk<0.5 and coapp_sk>-0.5:
  print('Fairly skewed data')       
elif coapp_sk>=0.5 and coapp_sk<=1 or coapp_sk<=-0.5 and coapp_sk>=-1:
    print('Moderately skewed data')
```

```python
loan_sk=data['LoanAmount'].skew()
print(loan_sk)
if(loan_sk>1 or loan_sk<-1):
    print("Highly skewed data")
elif loan_sk<0.5 and loan_sk>-0.5:
    print("Fairly skewed data")
elif loan_sk>=0.5 and loan_sk<=1 or loan_sk<=-0.5 and loan_sk>=-1:
    print("Moderately skewed data")
```

```python
loan_am_skw=data['Loan_Amount_Term'].skew()
print(loan_am_skw)
if(loan_am_skw>1 or loan_am_skw<-1):        
    print("Highly skewed data")
elif loan_am_skw<0.5 and loan_am_skw>-0.5:
    print("Fairly skewed data")
elif loan_am_skw>=0.5 and loan_am_skw<=1 or loan_am_skw<=-0.5 and loan_am_skw>=-1:
    print("Moderately skewed data")
```

```python
data['Gender'].fillna(data['Gender'].mode()[0],inplace=True)
data
```

```python
data['Dependents'].fillna(data['Dependents'].mode()[0],inplace=True)
```

```python
data['Married'].fillna(data['Married'].mode()[0],inplace=True)
data['Education'].fillna(data['Education'].mode()[0],inplace=True)
# data['Self-Employed'].fillna(data['Self-Employed'].mode()[0],inplace=True)
data['ApplicantIncome'].fillna(data['ApplicantIncome'].median(),inplace=True)
data['CoapplicantIncome'].fillna(data['CoapplicantIncome'].median(),inplace=True)
data['LoanAmount'].fillna(data['LoanAmount'].median(),inplace=True)
data['Loan_Amount_Term'].fillna(data['Loan_Amount_Term'].median(),inplace=True)
# data['Credit_History'].fillna(data['Credit_History'].mean()[0],inplace=True)
```

```python
df3=pd.get_dummies(data['Education'],dtype=int,drop_first=True,prefix='Edu')
data=pd.concat([data,df3],axis=1).drop('Education',axis=1)
df4=pd.get_dummies(data['Self_Employed'],dtype=int,drop_first=True,prefix='SE')
data=pd.concat([data,df4],axis=1).drop('Self_Employed',axis=1)
```
